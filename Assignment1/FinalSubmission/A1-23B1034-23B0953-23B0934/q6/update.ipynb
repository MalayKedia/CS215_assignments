```python
import math
def UpdateMean(OldMean,NewDataValue,n,A):
    NewMean=(n*OldMean+NewDataValue)/(n+1)   #Derivation in the report
    return NewMean
```

```python
def UpdateMedian(OldMedian,NewDataValue,n,A):
    A.append(NewDataValue)
    A.sort()
    if(n%2==1):
        NewMedian=(OldMedian+A[((n-1)//2)+1])/2
    else:
        NewMedian=A[(n)//2]
    return NewMedian
```

```python
def UpdateStd(OldMean,OldStd,NewMean,NewDataValue,n,A):
    NewStd=math.sqrt(((n-1)*pow(OldStd,2)+n*pow(OldMean,2)+pow(NewDataValue,2)-(n+1)*pow(NewMean,2))/n) #Derivation in the report
    return NewStd
```

```python
#For testing
A=[1,2,3,4,5,6,7]
n=7
OldMean=4
OldMedian=4
NewDataValue=8
NewMean=4.5
OldStd=math.sqrt(14/3)
print(UpdateMean(OldMean,NewDataValue,n,A))
print(UpdateMedian(OldMedian,NewDataValue,n,A))
print(A)
print(UpdateStd(OldMean,OldStd,NewMean,NewDataValue,n,A))
```
